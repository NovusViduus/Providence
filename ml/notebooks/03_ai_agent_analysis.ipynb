# AI Agent Detection — Behavioral Analysis

Feature distribution comparison: HUMAN vs AI_AGENT sessions.
Focus on timing features where the hypothesis predicts clearest separation.
- Side-by-side histograms per feature
- t-SNE/PCA 2D visualization
- Correlation matrices per class

```python
import sys
sys.path.insert(0, '..')

from src.data.synthetic_loader import build_dataset
from src.features.behavioral import AI_DETECTION_FEATURES
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Update paths to your data directories
X, y = build_dataset('data/human', 'data/ai_agent')
df = X.copy()
df['label'] = y.values
print(f'Dataset: {len(df)} samples')
print(y.value_counts())
```

```python
# Timing feature distributions — the key hypothesis
timing_features = ['inter_attempt_mean', 'inter_attempt_std', 'inter_attempt_cv', 'session_duration']
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, feat in zip(axes.flat, timing_features):
    for label in ['HUMAN', 'AI_AGENT']:
        subset = df[df['label'] == label][feat]
        ax.hist(subset, bins=30, alpha=0.6, label=label)
    ax.set_title(feat)
    ax.legend()
plt.tight_layout()
plt.show()
```

```python
# t-SNE visualization
from sklearn.manifold import TSNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_2d = tsne.fit_transform(X.values)
plt.figure(figsize=(10, 8))
for label, color in [('HUMAN', 'blue'), ('AI_AGENT', 'red')]:
    mask = y == label
    plt.scatter(X_2d[mask, 0], X_2d[mask, 1], c=color, alpha=0.5, label=label, s=20)
plt.legend()
plt.title('t-SNE: HUMAN vs AI_AGENT')
plt.show()
```
